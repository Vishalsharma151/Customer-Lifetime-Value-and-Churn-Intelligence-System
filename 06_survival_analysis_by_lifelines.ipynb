```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path
from lifelines import KaplanMeierFitter, CoxPHFitter
from lifelines.statistics import logrank_test
```

### Load the dataset

```python
# Use the same source dataset as Notebook 03. This loads the original customer records rather than reusing the model's one-hot-encoded features.

data_path = Path(r"C:\CHURNIQ\SQL\psql telco_customers_data.csv")

if not data_path.exists():
    raise FileNotFoundError(
        f"Dataset not found: {data_path}\n"
        "Update data_path to the actual CSV location."
    )

df = pd.read_csv(data_path)

# Standardize column names
df.columns = df.columns.str.strip().str.lower()

print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist())
df.head()
```

### Prepare tenure and churn-event variables

```python
# Here, churn_event = 1 means the customer churned; 0 means the customer had not churned at the observation point.

required_cols = ["tenure", "churn"]

missing_cols = [
    col for col in required_cols if col not in df.columns
]

if missing_cols:
    raise ValueError(f"Missing required columns: {missing_cols}")

survival_df = df.copy()

survival_df["tenure"] = pd.to_numeric(
    survival_df["tenure"], errors="coerce"
)

survival_df["churn"] = (
    survival_df["churn"].astype(str).str.strip().str.lower()
)

survival_df["churn_event"] = survival_df["churn"].map({
    "yes": 1,
    "no": 0
})

# Keep valid tenure and event observations
survival_df = survival_df.dropna(
    subset=["tenure", "churn_event"]
)

survival_df = survival_df[survival_df["tenure"] > 0].copy()
survival_df["churn_event"] = survival_df["churn_event"].astype(int)

print("Records for survival analysis:", len(survival_df))
print("\nChurn event counts:")
print(survival_df["churn_event"].value_counts())
```

### Kaplan–Meier overall retention curve

```python
# The Kaplan–Meier estimator estimates the probability of remaining churn-free over time under the survival-analysis assumptions.

kmf = KaplanMeierFitter()

kmf.fit(
    durations=survival_df["tenure"],
    event_observed=survival_df["churn_event"],
    label="All customers"
)

plt.figure(figsize=(9, 6))
kmf.plot_survival_function(ci_show=True)

plt.title("Kaplan–Meier Customer Retention Curve")
plt.xlabel("Tenure (months)")
plt.ylabel("Estimated churn-free survival probability")
plt.ylim(0, 1.05)
plt.grid(alpha=0.3)
plt.tight_layout()

output_dir = Path("../reports/survival")
output_dir.mkdir(parents=True, exist_ok=True)

plt.savefig(
    output_dir / "kaplan_meier_overall.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()

# Save estimated survival curve
km_curve = kmf.survival_function_.reset_index()
km_curve.to_csv(
    output_dir / "kaplan_meier_survival_curve.csv",
    index=False
)

print("Median estimated survival time:", kmf.median_survival_time_)
```

### Compare retention by contract type

```python
# This cell compares contract groups if the dataset contains the contract column.

if "contract" not in survival_df.columns:
    raise ValueError("The dataset does not contain a 'contract' column.")

plt.figure(figsize=(10, 6))

contract_kmf = {}

for contract_type, group in survival_df.groupby("contract"):
    km = KaplanMeierFitter()
    km.fit(
        durations=group["tenure"],
        event_observed=group["churn_event"],
        label=str(contract_type)
    )
    km.plot_survival_function(ci_show=False)
    contract_kmf[contract_type] = km

plt.title("Customer Retention by Contract Type")
plt.xlabel("Tenure (months)")
plt.ylabel("Estimated churn-free survival probability")
plt.ylim(0, 1.05)
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    output_dir / "kaplan_meier_by_contract.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

### Log-rank test for contract groups

```python
#The log-rank test assesses whether survival curves differ between two groups. We'll compare the two largest contract groups if at least two groups are available.

contract_counts = survival_df["contract"].value_counts()

if len(contract_counts) >= 2:
    group_names = contract_counts.index[:2]

    group_a = survival_df[
        survival_df["contract"] == group_names[0]
    ]
    group_b = survival_df[
        survival_df["contract"] == group_names[1]
    ]

    test_result = logrank_test(
        group_a["tenure"],
        group_b["tenure"],
        event_observed_A=group_a["churn_event"],
        event_observed_B=group_b["churn_event"]
    )

    logrank_results = pd.DataFrame([{
        "group_a": group_names[0],
        "group_b": group_names[1],
        "test_statistic": test_result.test_statistic,
        "p_value": test_result.p_value
    }])

    display(logrank_results)

    logrank_results.to_csv(
        output_dir / "contract_logrank_test.csv",
        index=False
    )
else:
    print("At least two contract groups are needed.")
```

### Prepare data for the Cox Proportional Hazards model

```python
# We'll use a small, interpretable set of customer characteristics. Only columns present in your dataset will be selected.

candidate_features = [
    "contract",
    "monthlycharges",
    "seniorcitizen",
    "internetservice",
    "paperlessbilling",
    "paymentmethod"
]

cox_features = [
    col for col in candidate_features
    if col in survival_df.columns
]

print("Cox model features:", cox_features)

if not cox_features:
    raise ValueError(
        "No candidate Cox model features were found."
    )

cox_df = survival_df[
    ["tenure", "churn_event"] + cox_features
].copy()

# Convert categorical features to indicator columns
cox_df = pd.get_dummies(
    cox_df,
    columns=[
        col for col in cox_features
        if (
            cox_df[col].dtype == "object"
            or str(cox_df[col].dtype) == "category"
            or cox_df[col].dtype == "bool"
        )
    ],
    drop_first=True,
    dtype=int
)

# Convert remaining values to numeric and remove invalid rows
cox_df = cox_df.apply(pd.to_numeric, errors="coerce")
cox_df = cox_df.replace([np.inf, -np.inf], np.nan).dropna()

# Remove constant columns that cannot help estimate coefficients
constant_cols = [
    col for col in cox_df.columns
    if col not in ["tenure", "churn_event"]
    and cox_df[col].nunique() <= 1
]

cox_df = cox_df.drop(columns=constant_cols)

print("Cox model dataset shape:", cox_df.shape)
cox_df.head()
```

### Fit the Cox model

```python
#Hazard ratio above 1: associated with a higher instantaneous churn hazard.
#Hazard ratio below 1: associated with a lower instantaneous churn hazard.
#Confidence interval and p-value: help quantify uncertainty in the estimated association.

cph = CoxPHFitter(penalizer=0.01)

cph.fit(
    cox_df,
    duration_col="tenure",
    event_col="churn_event"
)

cph.print_summary()
```

### Plot and save hazard ratios

```python
hazard_results = cph.summary.reset_index()

# Rename the first column for clarity
hazard_results = hazard_results.rename(
    columns={hazard_results.columns[0]: "feature"}
)

display(hazard_results)

hazard_results.to_csv(
    output_dir / "cox_hazard_ratios.csv",
    index=False
)

# Plot hazard ratios with confidence intervals
cph.plot()
plt.title("Cox Proportional Hazards Model Coefficients")
plt.tight_layout()

plt.savefig(
    output_dir / "cox_coefficients.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

### Check the proportional hazards assumption

```python
cph.check_assumptions(
    cox_df,
    p_value_threshold=0.05,
    show_plots=True
)
```

### Save a summary of the analysis

```python
survival_summary = pd.DataFrame([{
    "customers_analyzed": len(survival_df),
    "observed_churn_events": int(survival_df["churn_event"].sum()),
    "non_churned_at_observation": int(
        (survival_df["churn_event"] == 0).sum()
    ),
    "median_tenure_months": survival_df["tenure"].median(),
    "kaplan_meier_median_survival_months": kmf.median_survival_time_,
    "cox_concordance_index": cph.concordance_index_
}])

display(survival_summary)

survival_summary.to_csv(
    output_dir / "survival_analysis_summary.csv",
    index=False
)

print("Survival analysis results saved to:", output_dir.resolve())
```
